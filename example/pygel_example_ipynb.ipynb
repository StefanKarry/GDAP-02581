# Jupter notebook version of PyGEL example

## Imports

```python
from pygel3d import hmesh, jupyter_display as jd
from numpy import zeros
from plotly.io import renderers
renderers.default = "notebook_connected"
jd.set_export_mode()
```

## Example

```python
m = hmesh.load('bunny.obj')
hmesh.quadric_simplify(m, 0.05)
```

```python
jd.display(m)
```

```python
areas = zeros(m.no_allocated_vertices()) # Set up empty array to hold areas

# Iterating over all vertices
for v in m.vertices():
    # For each vertex, visit all faces incident to it
    a = 0
    for f in m.circulate_vertex(v, 'f'): # Circulare around vertex v and get all incident faces
        # Check if face is valid:
        try:
            a += m.area(f) # Add the area of each face to the total
        except OSError:
            pass

    areas[v] = a
```

```python
# Display the mesh, but using the areas as data
jd.display(m, data=areas)
```
